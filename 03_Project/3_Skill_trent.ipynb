## How are in- demand skills trending for Data Analysts?

 ### Methology
 
 1. Aggregate skill counts monthly
 2. Re-analyze based on percentage of total jobs
 3. Plot the monthly skill demand
 

```python
## importing libraries 
import ast
import pandas as pd
import seaborn as sns 
from datasets import load_dataset
import matplotlib.pyplot as plt 

## loading data 
dataset= load_dataset('lukebarousse/data_jobs')
df= dataset['train'].to_pandas()

## data cleaning 
df['job_posted_date']=pd.to_datetime(df['job_posted_date'])
df['job_skills']=df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df_DA_US=df[(df['job_title_short']=='Data Analyst') & (df['job_country']=='United States')].copy()

df_DA_US['job_posted_month']=df_DA_US['job_posted_date'].dt.month

df_DA_US_exploded=df_DA_US.explode('job_skills')
```

```python
df_DA_US_pivot=df_DA_US_exploded.pivot_table(index='job_posted_month', columns='job_skills', aggfunc='size', fill_value=0)

df_DA_US_pivot.loc['Total']=df_DA_US_pivot.sum()

df_DA_US_pivot= df_DA_US_pivot[df_DA_US_pivot.loc['Total'].sort_values(ascending=False).index]

df_DA_US_pivot=df_DA_US_pivot.drop('Total')

df_DA_US_pivot
```

```python
DA_total=df_DA_US.groupby('job_posted_month').size()
```

```python
df_DA_US_per=df_DA_US_pivot.div(DA_total/100,axis=0)
```

```python
## change month number to month name
df_DA_US_per = df_DA_US_per.reset_index()
df_DA_US_per['job_posted_months'] = df_DA_US_per['job_posted_month'].apply(lambda x: pd.to_datetime(str(x), format='%m').strftime('%b'))
df_DA_US_per = df_DA_US_per.set_index('job_posted_months')  
df_DA_US_per= df_DA_US_per.drop(columns='job_posted_month')

df_DA_US_per
```

```python
df_plot=df_DA_US_per.iloc[:, :5]
sns.lineplot(data=df_plot,dashes=False,palette='tab10')
sns.set_theme(style='ticks')
plt.title('Top 5 In-Demand Skills for Data Analysts in the US', fontsize=16)
plt.ylabel('Percentage of Job Postings (%)', fontsize=12)
plt.xlabel('Month', fontsize=12)


plt.show()
```
