```python
## importing libraries 
import ast
import pandas as pd 
from datasets import load_dataset
import matplotlib.pyplot as plt 

## loading data 
dataset= load_dataset('lukebarousse/data_jobs')
df= dataset['train'].to_pandas()

## data cleaning 
df['job_posted_date']=pd.to_datetime(df['job_posted_date'])
df['job_skills']=df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)

## filter for data analyst data 
df_DA=df[df['job_title_short']=='Data Analyst'].copy()
```

```python
## create a new column for month numbers 

df_DA['job_posted_month_no']=df_DA['job_posted_date'].dt.month

## explode the job skill coloum and pivot 
df_DA_explode=df_DA.explode('job_skills')
df_DA_pivot=df_DA_explode.pivot_table(index='job_posted_month_no',columns='job_skills',aggfunc='size',fill_value=0)

## sort the skills by count 
df_DA_pivot.loc['Total'] = df_DA_pivot.sum()
df_DA_pivot=df_DA_pivot[df_DA_pivot.loc['Total'].sort_values(ascending=False).index]
df_DA_pivot=df_DA_pivot.drop('Total')

## use month name for plotting 
df_DA_pivot=df_DA_pivot.reset_index()
df_DA_pivot['job_posted_month']=df_DA_pivot['job_posted_month_no'].apply(lambda x: pd.to_datetime(x,format='%m').strftime('%b'))
df_DA_pivot=df_DA_pivot.set_index('job_posted_month')
df_DA_pivot=df_DA_pivot.drop(columns='job_posted_month_no')

```

```python
## get the top 5 skills 
df_DA_pivot.iloc[:, :5].plot(
    kind='line',
    linewidth=4,
    linestyle=':',
    colormap='viridis',
    marker='o',
    markersize='5',
    figsize=(10,5)
    )
plt.title('Top 5 Skills for Data Analyst per month')
plt.ylabel('count')
plt.xlabel("")

plt.show()
```

```python
## calculate the average salary and count of job posting per skills 
skill_stats=df_DA_explode.groupby('job_skills').agg(
    median_salary=('salary_year_avg','median'),
    skill_count=('job_skills','count')
)

## limit to the top  skills 
skill_count=20
skill_stats=skill_stats.sort_values(by='skill_count',ascending=False).head(skill_count)


```

```python
from adjustText import adjust_text
```

```python
## scatter plot 
skill_stats.plot(kind='scatter',x='skill_count',y='median_salary')

## adding text annotations 
texts=[]
for i , txt in enumerate(skill_stats.index):
 texts.append(plt.text(skill_stats['skill_count'].iloc[i],skill_stats['median_salary'].iloc[i],txt))
adjust_text(texts, arrowprops=dict(arrowstyle='->', color='grey',lw=2))

## formatting the y axis for salary in 100k format 
ax=plt.gca()
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda y , pos: f'${int(y/1000)}K'))

## formatting the x axis with commas 
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x ,_: f'{int(x):,}'))

## set  axis labels and title 
plt.xlabel('Count of Job Posting')
plt.ylabel('Median Yearly Salary')
plt.title(f'Salary vs count of Job posting for Top {skill_count} Skills')

## adjust layout and display the plot 
plt.tight_layout()
plt.show()
```
